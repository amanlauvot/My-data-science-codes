```python
class employe:
    company="HP"
    def __init__(self,name,salary):
        self.name = name
        self.salary = salary 
    def printdetails(self):
        print(f"COMPANY IS {self.company} \nNAME OF EMPLOYE IS {self.name}\nSALARY OF THE EMPLOYE {self.salary}")

    @staticmethod
    def yeh():
       print(f"he is an good employe")

    @classmethod
    def printclass(cls):
        print(f"The company is {cls.company}")
e = employe("AMAN","50,000")
e.printdetails()
e.yeh()
e.printclass()
```
