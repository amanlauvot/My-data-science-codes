```python
#opening file
with open("finaldata.txt",encoding = 'utf-8') as f:
    data = f.read()
```

```python
#splitind paragraphed data
chunks = data.split("\n\n")
chunks = [c for c in chunks if len(c)>3]
```

```python
#spliting and making each user data in in different dictionary username,post_liked etc
def parse_chunk(chunk):
        chunk = chunk.strip()
        sep_chunk = chunk.split("\n")
        
        username = sep_chunk[0]
        # Converting post_liked,followers and following string to int
        post_liked = int(sep_chunk[1].split("post")[0].replace(",",""))
        followers = int(sep_chunk[2].split("followers")[0].replace(",","").replace("K","000").replace("M","000000").replace(".",""))
        following = int(sep_chunk[3].split("following")[0].replace(",","").replace("K","000").replace("M","000000").replace(".",""))
        name = sep_chunk[4]
        if(len(sep_chunk)>5):
            type_of_page = sep_chunk[5]
            bio = "\n".join(sep_chunk[6:])
        else:
            type_of_page = "unknown"
            bio = ""
        #print(username,post_liked,followers,following,name,type_of_page,bio ,sep = "\n")
    
        return{"user_name":username,"postliked":post_liked,"followers":followers,"following":following,"Name":name,"Type_of_page":type_of_page,"Bio":bio}

#print(parse_chunk(chunks[0]))
```

```python
#creating new list which contains all parsed chunks
all_chunks = []
for chunk in chunks:
    parsed_chunks = parse_chunk(chunk)
    all_chunks.append(parsed_chunks)
```
