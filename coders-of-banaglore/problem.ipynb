```python
#opening file
with open("initialdata.txt",encoding = 'utf-8') as f:
    data = f.read()
```

```python
#splitind paragraphed data
chunks = data.split("\n\n")
```

```python
chunks
```

```python
chunks[0]
```

```python
#spliting and making each user data in in different dictionary username,post_liked etc
def parse_chunk(chunk):
    chunk = chunk.strip()
    sep_chunk = chunk.split("\n")
    
    username = sep_chunk[0]
    post_liked = sep_chunk[1]
    followers = sep_chunk[2]
    following = sep_chunk[3]
    name = sep_chunk[4]
    type_of_page = sep_chunk[5]
    bio = "\n".join(sep_chunk[6:])
    print(username,post_liked,followers,following,name,type_of_page,bio ,sep = "\n")

parse_chunk(chunks[0])
```
